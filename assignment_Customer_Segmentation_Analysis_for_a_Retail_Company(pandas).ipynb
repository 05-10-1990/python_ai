# **Problem Statement: Customer Segmentation Analysis for a Retail Company**
Scenario:
You are a data analyst at a retail company. Your task is to analyze customer and sales data to identify high-value transactions, top-selling products, and segment customers based on their spending patterns. The goal is to provide actionable insights to improve customer targeting, retention strategies, and sales performance.

Download the dataset from the below link

[sales.xlsx](https://docs.google.com/spreadsheets/d/10qQs5F19MMFHmfco-IR3pWdKrCw8c07D/edit?usp=drive_link&ouid=110628496462120539988&rtpof=true&sd=true)

**Step 1: Read Customer and Sales Data**

Description: Import two datasets:
* Customer Data: Includes customer details such as customer_id, customer_name, customer_email, customer_age, and customer_location.
* Sales Data: Contains transactional data such as order_id, product_id, product_name, product_category, sales_amount, and order_date.

Expected Output:

image.png


```python
import pandas as pd

customer_df = pd.read_csv('content/customers.csv')

sale_df = pd.read_excel('content/sales.xlsx')

customer_df.head(10)
```

```python
sale_df.head(10)
```

**Step 2: Merge the Datasets on customer_id**

Description: Combine the customer and sales datasets based on customer_id to create a comprehensive view of customers and their purchases.

Expected Output:

image.png

```python
merge_df = pd.merge(customer_df,sale_df,on='customer_id')
merge_df
```

**Step 3: Filter for High-Value Transactions (sales_amount > 100)**

Description: Filter the merged dataset to focus only on transactions where the sales_amount exceeds 100. To prioritize analysis on significant transactions that contribute more to revenue

Expected output:

image.png

```python
# Step 3: Filter for high-value transactions (sales_amount > 100)
input_high_value = int(input('Enter Your value :'))

if len(merge_df) > 0:

    high_value_sale_df = merge_df[
        merge_df["sales_amount"] > input_high_value
    ]

    if len(high_value_sale_df) > 0:
        display(high_value_sale_df)
    else:
        print("No high-value transaction found")

else:
    print("No data found")
```

**Step 4: Sort High-Value Transactions by sales_amount**
Description: Sort the filtered dataset in descending order of sales_amount to identify top-selling products and high-value transactions.To determine which products or customers contribute the most to revenue.

Expected Output:

image.png

```python
# Step 4: Sort high-value transactions to identify top-selling products
top_selling_product_df = high_value_sale_df.sort_values(by='sales_amount',ascending=False)
top_selling_product_df
```

```python
top_selling_product_soft_alphabat_df = high_value_sale_df.sort_values(by='customer_name',ascending=True)
top_selling_product_soft_alphabat_df
```

**Step 5: Apply Segmentation Logic to Create Customer_Segment**
To categorize customers for better targeting and strategy formulation.
Description: Create a new column Customer_Segment based on the sales_amount :
* "High Value" if sales_amount >= 200.
* "Medium Value" if sales_amount is between 100 and 200.
* "Low Value" if sales_amount < 100.

Expected Output:

image.png


```python
# Step 5: Apply segmentation logic to 'sales_amount' to create 'Customer_Segment'
def segment_customer(sales):
  if sales >=200:
    return "High Value"
  elif sales >= 100:
    return "Medium Value"
  else:
    return "Low Value"

merge_df['Customer_Segment'] = merge_df['sales_amount'].apply(segment_customer)
merge_df
```

```python
# Save the final DataFrame to an Excel file
merge_df.to_excel("Customer_Segmentation.xlsx", index=False)
print("The final combined DataFrame has been saved to 'Customer_Segmentation.xlsx'.")
```
