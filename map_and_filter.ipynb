##**Real-World Applications of map() in Data Processing**

The map() function is commonly used in data processing tasks to apply transformations or calculations to large datasets.

**Extracting Names from a Database**

You have a list of customer records and need to extract just the names.

```python
customers = [{"name": "Alice", "age": 30}, {"name": "Bob", "age": 25}, {"name": "Eve", "age": 35}]

names = list(map(lambda customer :customer['name'],customers))
ages = list(map(lambda age : age['age'],customers))
print(f"name : {names}")
print(f"Age : {ages}")
```

```python
transactions = [200, -50, 150, -100, 500, -20]

# Use filter() to find only positive transactions (income)
positive_transaction = filter(lambda amount : amount > 0,transactions)

print(list(positive_transaction))
```
