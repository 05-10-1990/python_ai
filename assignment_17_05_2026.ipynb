Problem 4: Define a function with four parameters
Problem Statement: Create a function divide that accepts four numbers and returns the result of dividing the first by the sum of the remaining three numbers.

Expected Output: The division result.

```python
def division(a,b,c,d):
  return a / (b + c +d)

num1 = int(input("Please enter Input First number :"))
num2 = int(input("Please enter Input Second Number :"))
num3 = int(input("Please enter Input third Number :"))
num4 = int(input("Please enter Input fourth Number :"))
print("result of dividing the first by the sum of the remaining three numbers : ",division(num1,num2,num3,num4))
```

Problem 8 : Delete a Line by Index from a File
Problem Statement: Delete a specific line (e.g., the third line) from the file user_activity.txt. After deletion, update the file to reflect the changes using with statement.

Expected Output:

Updated content of the user_activity.txt file:

User logged in at 09:00 AM

User viewed product details at 09:15 AM

User logged out at 10:00 AM

User added item to wishlist at 10:30 AM

```python
# Read all lines from the file
with open("user_activity.txt", "r") as file:
    lines = file.readlines()

# Delete the third activity line (index 4 because of blank lines)
del lines[2]

# Write the updated content back to the file
with open("user_activity.txt", "w") as file:
    file.writelines(lines)

# Display the updated content
with open("user_activity.txt", "r") as file:
    print("Updated content of the user_activity.txt file:\n")
    print(file.read())
```

Problem 10 : Append a New User to the CSV File
Problem Statement: Append a new user's data (name, age, email) to the existing users.csv file and Display the updated content of the CSV file use with statement.

Expected Output:

Updated CSV file content:

John Doe,28,john.doe@example.com

Jane Smith,34,jane.smith@example.com

Bob Johnson,45,bob.johnson@example.com

Alice Green,29,alice.green@example.com

```python
import csv

data = [["Name", "Age", "Email"],["John Doe",28,"john.doe@example.com"],["Jane Smith",34,"jane.smith@example.com"],["Bob Johnson",45,"bob.johnson@example.com"]]

# Step 1 : open the csv file in write mode
with open('users.csv',mode='w',newline='') as file:
  # Create a CSV writer with pipe '|' as the delimiter
  writer = csv.writer(file) # csv.writer(file,delimiter='|')
  # Write the data rows to the file
  writer.writerows(data)
  print("Updated CSV file content:")

# Step 2 : Append data on users.csv file
# New data to add
new_row = ["Alice Green", 29, "alice.green@example.com"]
# Append new data to the CSV file
with open('users.csv', mode='a', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(new_row)

# Step 3 : Now, read and display the updated file to verify the change
# Print users.csv file data
with open("users.csv",mode="r") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```

Problem 11 : Delete a Row by Index from the CSV File
Problem Statement: Delete a user by index (e.g., the second user, "Jane Smith") from the users.csv file. Write the updated content back to the CSV file  Display the updated CSV file content use with statement.

Expected Output:

Updated CSV file content after deletion:

John Doe,28,john.doe@example.com

Jane Smith,34,jane.smith@example.com

Alice Green,29,alice.green@example.com


```python
import csv

# Step 1: Read all rows from the CSV file
with open("users.csv", mode="r") as file:
    rows = list(csv.reader(file))

# Step 2: Remove the row at index 2
rows.pop(2)

# Step 3: Write the updated rows back to the CSV file
with open("users.csv", mode="w", newline="") as file:
    csv.writer(file).writerows(rows)

# Step 4: Display updated content
with open("users.csv", mode="r") as file:
  print("Updated CSV file content after deletion:\n")
  for line in file:
    if line.strip() != 'Name,Age,Email':
      print(line.strip())

  print(file.read())
```

Problem 12: Read Specific User Data by Index from the CSV File
Problem Statement: Read and display the information of a specific user (e.g., the second user, "Jane Smith") from the users.csv file by using the index and with statement.

Specific user at index 3: ['Bob Johnson', '45', 'bob.johnson@example.com']

```python
import csv

data = [
    ['Name', 'Age', 'Email'],
    ['John Doe', 28, 'john.doe@example.com'],
    ['Jane Smith', 34, 'jane.smith@example.com'],
    ['Bob Johnson', 45, 'bob.johnson@example.com']
]

with open("userss.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("Uploaded CSV file Content:")

with open("userss.csv", mode="r") as file:
  for line in file:
    if line.strip() != 'Name,Age,Email':
      print(line.strip())
    print(file.read())


with open('userss.csv',mode="r") as file:
  rows = list(csv.reader(file))
print("Specific user at index 3:", rows[3])
```
