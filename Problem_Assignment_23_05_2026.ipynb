**4.Problem Statement:**
Write a lambda function to check if a given string is a palindrome. The lambda function should return True if the string is the same when read forwards and backwards, and False otherwise.

**Expected Output:**If the string is a palindrome, the output should be True, otherwise False.

**Example:**

Input: "racecar" Output: True
* "racecar" is a palindrome because if you read it from left to right or right to left, it’s the same.

Input: "hello" Output: False
* hello" → when read from left to right, it's "hello", but when read backward, it's "olleh". This is not a palindrome.

**Explanation:**A palindrome is a word or phrase that reads the same backward as forward, ignoring spaces, punctuation, and capitalization.


```python
is_palindrome = lambda text: text == text[::-1]

print(is_palindrome("racecar"))  # True
```

**6.Problem Statement:** Write a lambda function that takes a number as input and checks if it is divisible by both 2 and 3. The function should return True if the number is divisible by both, otherwise return False.

**Expected Output:**
The output should be True if the number is divisible by both 2 and 3, otherwise False.

**Example:**
* Input: 6
* Output: True
* (6 is divisible by both 2 and 3)

* Input: 8
* Output: False
* (8 is divisible by 2 but not by 3)

```python
check_divisible = lambda x: x % 2 == 0 and x % 3 == 0

print(check_divisible(8))
```

**7.Problem Statement:**
Write a program that uses the map() function along with a lambda function to multiply each element in the given list by 2. The output should be a new list with each element multiplied by 2.

**Expected Output:**
The output should be a list where each number is multiplied by 2.

Example:

* Input: [1, 2, 3, 4, 5]
* Output: [2, 4, 6, 8, 10]

* Input: [10, 20, 30]

**8.Problem Statement:**
Write a program that uses the map() function along with a lambda function to return the length of each word in the given list of words. The output should be a list containing the length of each word.

**Expected Output:**
The output should be a list of integers where each element is the length of the corresponding word in the input list.

**Example:**
* Input: ["apple", "banana", "cat"]
* Output: [5, 6, 3]

* Input: ["hello", "world"]
* Output: [5, 5]

```python
words = ["apple", "banana", "cat"]

result = list(map(lambda word: len(word), words))

print(result)
```

**9.Problem Statement:**
Write a program that uses the map() function along with a lambda function to convert a list of strings to integers. The output should be a list containing the integer representation of each string in the input list.

**Expected Output:**
The output should be a list of integers corresponding to each string in the input list.

**Example:**
* Input: ["1", "2", "3", "4"]
* Output: [1, 2, 3, 4]

* Input: ["10", "20", "30"]
* Output: [10, 20, 30]

```python
strings = ["1", "2", "3", "4"]

result = list(map(lambda x: int(x), strings))

print(result)
```

**10.Problem Statement:**
Write a Python program that uses the map() function and a lambda function to add corresponding elements from two lists of equal length. The program should return a list of sums where each element is the sum of the corresponding elements from both lists.

Expected Output:The output is a list of sums calculated for each pair of corresponding elements from list1 and list2:

1 + 5 = 6

2 + 6 = 8

3 + 7 = 10

Thus, the program outputs [6, 8, 10].


```python
list1 = [1, 2, 3]
list2 = [5, 6, 7]

result = list(map(lambda x, y: x + y, list1, list2))

print(result)
```
