**Step 1: Create Two Tables**

```python
# Sqlite module - functions are there

import sqlite3

# Connect to the database
connection = sqlite3.connect('bookstore.db')
cursor = connection.cursor()

# Create Books table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Books (
  title TEXT,
  author text,
  price INTEGER,
  stock INTEGER
  )
''')

# Create Sales table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Sales (
  book_title TEXT,
  quantity INTEGER,
  sale_date DATE
)
''')

print("Tables created successfully.")
connection.commit()
```

**Step 3: List All Books in the Store**

```python
cursor.execute("SELECT * FROM books")
books = cursor.fetchall()

print('books Available in store:')
for book in books:
  print(f"Title : {book[0]}, Author : {book[1]}, Price : {book[2]}, Stock : {book[3]}")
```

**Step 4: Check for Low Stock**

```python
cursor.execute('SELECT title, stock FROM books WHERE stock < 3')
rows = cursor.fetchall()

for row in rows:
  print(f"{row[0]} : {row[1]} copies left")
```

**Step 5: Find Affordable Books for Promotions**

```python
cursor.execute('SELECT title, price FROM books WHERE price < 10')
rows = cursor.fetchall()

print('Books Priced Below $10:')
for row in rows:
  print(f"{row[0]} : ${row[1]}")
```

**Step 6: Identify Unsold Books**



```python
cursor.execute('SELECT title FROM books WHERE title NOT IN (SELECT book_title FROM sales)');
rows = cursor.fetchall()

print("\nBooks That Haven’t Been Sold:")
for row in rows:
    print(row[0])

connection.commit()
connection.close()
```
