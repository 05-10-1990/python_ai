1. Sales Trends Over Time (Line Plot)

Goal: Understand how daily sales revenue changes over the year.

Solution: A line plot will help us visualize daily sales revenue trends. This can reveal patterns like consistent growth, seasonal peaks, or dips in sales.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the dataset
sales_data = pd.read_csv('content/sales_data.csv')

# Group sales revenue by date
sales_data['Date'] = pd.to_datetime(sales_data['Date']) # Convert to datetime
sales_trends = sales_data.groupby('Date')['Sales Revenue'].sum()

# Plot the line chart with improved formatting
plt.figure(figsize=(12,6))
plt.plot(sales_trends.index,sales_trends.values,marker='o',color='green',linewidth=2)

# Improve X-axis
plt.title('Sales Revenue Over Time', fontsize=16)
plt.xlabel('Date', fontsize=14)
plt.ylabel('Sales Revenue', fontsize=14)
plt.xticks(ticks=sales_trends.index[::30], rotation=45)
plt.grid(alpha=0.5)

# Show the plot
plt.tight_layout()
plt.show()
```

2. Product Category Insights (Count Plot)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the count of sales for each product category
plt.figure(figsize=(10,6))
sns.countplot(data=sales_data,x='Product Category',palette='rocket')
plt.title('Frequency of Sales by Product Category',fontsize=16)
plt.xlabel('Product Category', fontsize=14)
plt.ylabel('Count of Sales', fontsize=14)
plt.xticks(rotation=45)
plt.show()
```

3. Regional Performance (Bar Plot)

```python
regional_sales = sales_data.groupby('Region')['Sales Revenue'].sum().reset_index()

regional_sales
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.barplot(data=regional_sales, x='Region', y='Sales Revenue', palette='coolwarm')
plt.title('Sales Revenue by Region', fontsize=16)
plt.xlabel('Region', fontsize=14)
plt.ylabel('Sales Revenue', fontsize=14)
plt.show()
```

4. Profit Variability (Box Plot)

```python
# Plot a box plot for profit by region

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sales_data = pd.read_csv('content/sales_data.csv')

plt.figure(figsize=(10,6))

sns.boxplot(data=sales_data, x='Region', y='Profit',hue='Region', palette='magma',legend=False)
plt.title('Profit Variability by Region',fontsize=16)
plt.xlabel('Region',fontsize=12)
plt.ylabel('Profit',fontsize=12)
plt.show()
```

5. Revenue by Product Category (Pie Chart)

```python
import matplotlib.pyplot as plt
import pandas as pd

sales_data = pd.read_csv('content/sales_data.csv')

# Calculate total revenue by product category
Category_Revenue = sales_data.groupby('Product Category')['Sales Revenue'].sum()

print(Category_Revenue.index)
# Create a pie chart
plt.pie(Category_Revenue.values,labels=Category_Revenue.index,autopct="%1.1f%%",startangle=90,)

plt.title('Revenue Contribution by Product Category')

# Show the plot
plt.show()
```
