**What are Lambda Functions?**

Lambda functions are small, anonymous functions in Python that are defined using the lambda keyword. They are used to write quick, one-line functions without giving them a name.

**Why Use Lambda Functions?**

* Concise and Readable: They reduce the need for defining full functions when performing simple tasks.
* Quick to Write: Useful for small tasks, especially when used inline with other functions.
* No Namespace Pollution: They remain unnamed and don’t clutter your codebase.

-- Syntax and Structure of a Lambda Function

**The basic syntax is:**

lambda arguments: expression

* lambda is the keyword
* arguments are the inputs to the function.
* expression is the operation or computation performed on the arguments.


## **Comparing def and lambda for Doubling a Number**
Here’s how you can use both def and lambda to create functions that double a number:

**Using def Function**

```python
# Define a function using def

def double(x):
  return x * 2

# Call Function
result = double(4)
print(result)
```

**Using lambda Function**

Example:You are building a calculator app and want to quickly double a number.

* Input: The number 4.

* Output: The result is 8.
------------------------------------------------------
* lambda → keyword used to create a small, anonymous (no-name) function
* x → input (parameter)

You are developing a grading system system where you need to calculate the total marks obtained by a student across three subjects: Math, Science, and English.

**When to Use Lambda Functions Instead of Regular/defining Functions**
* Use Lambda Functions:
When you need a simple, one-line function for a quick task, especially if it's not going to be reused elsewhere in your code.

* Use Regular/defining Functions:
When the task is more complex, involves multiple operations, or will be used multiple times across your code. Regular functions are better for maintaining readability and making your code more structured.

## **Using def Function for Classifying Numbers**

**Example:**Classifying Numbers as Positive or Negative
Here's how you can achieve the same classification using a def function:



```python
def classify(x):
  if x > 0:
    return "Positive"
  else:
    return "Negative"

#call the function
print(classify(4)) # Positive
print(classify(-5)) # Negative
```

**Example**:Classifying Numbers as Positive or Negative A lambda function that classifies a number as "Positive" or "Negative" based on its value.


Here's how you can achieve the same classification using a lambda
function:

```python
classify = lambda x : "Positive" if x > 0 else "Negative"

print(classify(-5)) # Negative
print(classify(3)) # Positive
```

**Using def Function to Check Even or Odd Numbers**

Here's how you can implement the same functionality using a def function:

```python
def check_even_odd(x):
  if x % 2 == 0:
    return 'Even'
  else:
    return 'Odd'

print(check_even_odd(4)) # Even
print(check_even_odd(3)) # Odd
```

**Example**:Checking Even or Odd A lambda function that checks whether a number is even or odd.

```python
check_even_odd = lambda x : "Even" if x % 2 == 0 else "Odd"

print(check_even_odd(4)) # Even
print(check_even_odd(7)) # Odd
```

**What is map() in Python?**

The map() function in Python applies a specific operation to each item in a collection, like a list, and returns a new collection with the results.

For example, to double each number in a list, map() can quickly apply the operation without needing a loop. It takes a function (the operation) and the collection, processes each item, and returns the updated values.

**Purpose and Benefits of Using map()**

* Efficient Transformation: Easily applies the same operation to all items in a collection without a loop.
* Concise Code: Makes code shorter and cleaner compared to using loops.
* Improves Readability: Clearly shows the intent of applying a function to all items.
* Supports Functional Programming: Works well with functions and lambda expressions for quick data processing.

**Basic Syntax of map()**

-- The map() function follows this syntax:

**map(function, collection)**

* function: The operation to apply to each item (can be a built-in, user-defined, or lambda function).
* collection: The items on which the function is applied (like a list or tuple)

**Example: Double Each Number in a List**

Imagine you have a list of numbers, and you want to create a new list where each number is doubled. Instead of writing a loop to achieve this, you can use the map() function for a simpler and cleaner solution.

```python
numbers = [1,2,3,4,5]
double = map(lambda x : x **2,numbers)
print(double)
print(list(double))
```

```python
double = list(map(lambda x : x**2,numbers))
print(double)
```

This means:

* Each number in the list (1, 2, 3, 4) is passed into the lambda function one by one.
* The result of x * 2 is calculated for each number (2, 4, 6, 8).
* A new map object is created to hold the transformed results.
* Converting to a List: Since map() returns a special map object, you need to convert it into a list to see the results:

* Final Result
The output [2, 4, 6, 8] is a new list where each number from the original list has been doubled.

**Doubling Numbers Without Using map()**

Here's how you can achieve the same result without using the map() function:

```python
# list of numbers
numbers = [1, 2, 3, 4]

# Double each number using a loop
doubled = []
for x in numbers:
  doubled.append(x**2)

print(doubled)
```

map() makes code shorter and easier to read by eliminating the need to manually create and append to a list in a loop.

**Example: Add Shipping Fee to Product Prices**

You want to add a fixed shipping fee (e.g., $5) to each price.

```python
# List of base product prices
product_prices = [20, 45, 70, 100]

# Use map() with a lambda to add a shipping fee
final_price = list(map(lambda x : x + 5,product_prices))

print("Base Price = ",product_prices)
print("Product Price = ",final_price)
```

## **Using map() with multiple lists**

**Example: Combining First and Last Names**

Imagine you have two lists, one with first names and another with last names. Your goal is to combine them to create a list of full names.

```python
# Two lists: first names and last names
first_names = ["John", "Jane", "Sam"]
last_names = ["Doe", "Smith", "Brown"]

# Using map with both lists
full_name = list(map(lambda first, last : first + " " + last,first_names,last_names))
print(full_name)
```

```python
# Two lists: first names and last names
first_name = ["Ropil", "Vijay", "Sita"]
last_name = ["Jain", "Sharma", "Kumari"]
list(map(lambda first,last : first + " " + last,first_name,last_name))
```

## **Using map() with multiple tuple**

**Guest List for a Formal Event**

You are preparing a guest tuple for a formal event, like a seminar or gala. To ensure proper respect, you combine titles (e.g., Dr., Mr., Ms.) with last names to create full names. This ensures invitations, name tags, and documents look professional.

```python
title = ("Dr.", "Mr.", "Ms.")
name = ("Smith", "Johnson", "Taylor")

# Combine titles and names
full_names = tuple(map(lambda title,name : f"{title} {name}",title,name))

# Print the full name list
print(full_names)
```

```python
customers = [{"name": "Alice", "age": 30}, {"name": "Bob", "age": 25}, {"name": "Eve", "age": 35}]

names = list(map(lambda customer :customer['name'],customers))
ages = list(map(lambda age : age['age'],customers))
print(f"name : {names}")
print(f"Age : {ages}")
```

```python
transactions = [200, -50, 150, -100, 500, -20]

# Use filter() to find only positive transactions (income)
positive_transaction = filter(lambda amount : amount > 0,transactions)

print(list(positive_transaction))
```
